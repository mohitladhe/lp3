# Practical: K-Means Clustering on Sales Data

## Objective
Implement K-Means clustering on the `sales_data_sample.csv` dataset and determine the optimal number of clusters using the Elbow Method.

## Theory
K-Means is an unsupervised machine learning algorithm used to divide data into `K` clusters based on similarity.

### Steps
1. Read the sales dataset.
2. Select relevant numerical features.
3. Handle missing values.
4. Normalize the data.
5. Apply the Elbow Method to find the suitable value of K.
6. Apply K-Means clustering.
7. Visualize the clusters.
8. Display cluster-wise sales information.

The Elbow Method selects K by plotting the Within-Cluster Sum of Squares (WCSS) against the number of clusters. The point where the decrease in WCSS starts becoming gradual is considered the elbow.


### Theory
This cell imports Pandas for data handling, Matplotlib and Seaborn for graphs, and KMeans and StandardScaler from Scikit-learn.

### Output Explanation
The message confirms that all required libraries were imported successfully.

```python
# Step 1: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

print("Libraries imported successfully!")
```

### Theory
This cell reads the sales CSV file into a DataFrame. The latin1 encoding helps read special characters that may be present in the file.

### Output Explanation
The shape shows the number of rows and columns. The table displays the first five sales records.

```python
# Step 2: Read Dataset
# Keep sales_data_sample.csv in the same folder as this notebook.
df = pd.read_csv("sales_data_sample.csv", encoding="latin1")

print("Dataset Shape:", df.shape)
df.head()
```

### Theory
This cell checks the dataset columns, data types, and missing values. This helps us understand the data before clustering.

### Output Explanation
The information shows column types and record counts. The missing-values list shows how many empty values are present in each column.

```python
# Step 3: Explore Dataset
df.info()

print("\nMissing Values:")
print(df.isnull().sum().sort_values(ascending=False).head(15))
```

### Theory
We select three numerical columns: quantity ordered, price per item, and sales amount. Rows with missing values in these columns are removed.

### Output Explanation
The table shows the selected columns. The shape shows how many complete records remain.

```python
# Step 4: Select Features
features = ["QUANTITYORDERED", "PRICEEACH", "SALES"]
data = df[features].dropna()

print(data.head())
print("\nShape after preprocessing:", data.shape)
```

### Theory
StandardScaler converts the selected features to a similar scale. This is important because K-Means uses distances to form clusters.

### Output Explanation
The message confirms that the sales features have been normalized.

```python
# Step 5: Normalize the Data
scaler = StandardScaler()
X = scaler.fit_transform(data)

print("Data normalization completed!")
```

## Step 6: Determine Number of Clusters Using Elbow Method

We calculate WCSS for K values from 1 to 10. The suitable K is selected from the elbow point of the curve.


### Theory
The Elbow Method runs K-Means for K values from 1 to 10 and stores WCSS for each value. WCSS measures how close records are to their cluster centers.

### Output Explanation
The graph plots K against WCSS. Choose the elbow point where the decrease starts becoming slower; this practical uses K = 3.

```python
# Step 6: Elbow Method
wcss = []

for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X)
    wcss.append(model.inertia_)

plt.plot(range(1, 11), wcss, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

### Choosing K

For this sales dataset, the elbow is typically around **K = 3** when using the selected features. The exact choice should be confirmed from the elbow plot generated above.

We therefore use **K = 3** for the final K-Means model.


### Theory
K-Means divides the sales records into three groups based on similarity. Each record receives a cluster label.

### Output Explanation
The output shows how many records belong to Cluster 0, Cluster 1, and Cluster 2.

```python
# Step 7: Apply K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
data["Cluster"] = kmeans.fit_predict(X)

print("Clustering completed!")
print(data["Cluster"].value_counts().sort_index())
```

### Theory
The scatter plot displays each sales record using quantity ordered and sales. The color represents its assigned cluster.

### Output Explanation
Points with the same color belong to the same cluster. Separated groups indicate different sales patterns.

```python
# Step 8: Visualize Clusters
sns.scatterplot(data=data, x="QUANTITYORDERED", y="SALES", hue="Cluster")
plt.title("K-Means Clusters")
plt.show()
```

### Theory
This cell calculates the average quantity, price, and sales for each cluster. It helps us understand the characteristics of each group.

### Output Explanation
The table shows the average values for each feature in Cluster 0, 1, and 2.

```python
# Step 9: Cluster Analysis
summary = data.groupby("Cluster")[features].mean()
print(summary)
```

### Theory
This cell counts the number of records in each cluster and displays the counts as a bar graph.

### Output Explanation
Each bar represents one cluster. A taller bar means that more sales records belong to that cluster.

```python
# Step 10: Cluster-wise Number of Records
counts = data["Cluster"].value_counts().sort_index()
counts.plot(kind="bar")

plt.xlabel("Cluster")
plt.ylabel("Number of Records")
plt.title("Records in Each Cluster")
plt.show()
```

## Conclusion

K-Means clustering was successfully implemented on the sales dataset. The Elbow Method was used to determine a suitable number of clusters. The sales records were standardized before clustering, and the resulting clusters were visualized and analyzed using their average quantity ordered, price, and sales.

### Viva Point
**K-Means** groups similar observations by minimizing the distance between observations and their cluster centroid. The **Elbow Method** helps select K by finding the point where additional clusters provide diminishing improvement in WCSS.
